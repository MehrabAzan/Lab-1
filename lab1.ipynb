```python
import numpy as np
```

```python
# Creating a array
variable_name = [1, 2, 3, 4, 5]
# Printing the array
print(variable_name)
# Printing the type of the array
print(type(variable_name))
```

```python
# Converting the array to a numpy array
new_variable = np.array(variable_name)
# Printing the numpy array
print(new_variable)
# Printing the type of the numpy array
print(type(new_variable))
```

```python
# Creating a 2D array
python_2d_array = [[9000, 2137, 21, 63, 9001], [67, 41, 666, 69, 34], [43, 61, 56, 420, 20]]
# Printing the 2D array
print(python_2d_array)
# Printing the type of the 2D array
print(type(python_2d_array))
```

```python
# Converting the 2D array to a numpy array
numpyArray = np.array(python_2d_array)
# Printing the numpy array
print(numpyArray)
# Printing the type of the numpy array
print(type(numpyArray))
```

```python
# Creating an array that contains the numbers from 0 to 14 with a step of 2
np.arange(0, 15 ,2)
```

```python
# Creating an array that contains the numbers from 0 to 15 that is reshaped into a 3 by 5 array
np.arange(0, 15).reshape(3, 5)
```

```python
# Creating an array that contains 15 zeros
np.zeros(15)
```

```python
# Creating an array that contains 15 zeros that is reshaped into a 3 by 5 array
np.zeros(15).reshape(3, 5)
```

```python
# Creating an array that contains 15 ones
np.ones(15)
```

```python
# Creating an array that contains 15 ones that is reshaped into a 3 by 5 array
np.ones(15).reshape(3, 5)
```

```python
# Creating an array that contains 9 zeros that is reshaped into a 3 by 3 array with the data type of int
np.zeros(9, dtype=int).reshape(3, 3)
```

```python
# Creating an array that contains the identity matrix of size 5 by 5
np.eye(5)
```

```python
# Creating an array that contains 20 random integers from 0 to 100 that is reshaped into a 4 by 5 array
A = np.random.randint(0, 100, 20).reshape(4, 5)
A
```

```python
# Printing the smallest and largest values in the array A, and their positions
print(f'The smallest value in A is {A.min()}, and is a located at position {A.argmin()}')
print(f'The largest value in A is {A.max()}, and is a located at position {A.argmax()}')
```
